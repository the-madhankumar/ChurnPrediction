```python
import pandas as pd

# 1. Loading Customer_chrun

file_path = "../dataset/telco_customer.csv"
df = pd.read_csv(file_path)
```

```python
df.head()
```

```python
# 2. Pringing all column names
df.columns.tolist()
```

```python
df["TotalCharges"] = pd.to_numeric(df["TotalCharges"], errors="coerce")
```

```python
# 3. Calculating null and count percent

row, columns = df.shape
def getPercent(value):
    return (value/row) * 100

for cols in df.columns.tolist():
    print(f"Column Name: {cols} , percent: {getPercent(df[cols].isnull().sum())}")
```

```python
df.info()
```

```python
# 4. Value Counts for categorical column

columns = ["gender", "Partner", "Dependents", "PhoneService", "InternetService", "Contract", "PaymentMethod", "Churn"]

for col in columns:
    print(f"========== {col} ==========")
    print("Unique Values :", ", ".join(df[col].unique().tolist()))
    print()
    print("Value Count")
    for value, count in df[col].value_counts().items():
        print(f"{value}: {count}")
    print()
```

```python
# 5. Churn rate percent
print("Churn")
for value, count in df["Churn"].value_counts(normalize = True).items():
        print(f"{value}: {round(count * 100, 2)}")
```

```python
# 6. TotalCharges NaN

# df["TotalCharges"] = pd.to_numeric(df["TotalCharges"], errors="coerce")

print(f"Total Charges Nan Values : {df["TotalCharges"].isnull().sum()}")
```

```python
cols = ["tenure", "MonthlyCharges", "TotalCharges"]
for col in cols:
    print(f"{"="*10} {col} {"="*10}")
    print(f"Mean : {df[col].mean()}")
    print(f"Min : {df[col].min()}")
    print(f"Max : {df[col].max()}")
    print()
```

```python
# 8. Profile Summary

print(f"{"="*25} Profiling Summary {"="*25}")
print(f"=> The Dataset has {df.shape[0]} rows and {df.shape[1]} columns.")
print(f"=> Total Charges converted to numeric and has {df["TotalCharges"].isnull().sum()} NaN values.")

churn_distribution = df["Churn"].value_counts(normalize=True).to_dict()
print(f"=> Churn Distribution: {", ".join([f"{k}: {round(v * 100,2)} %" for k, v in churn_distribution.items()])}, slightly imbalance.")
print(f"=> Most of the categorical values have limited unique values.")
print("=> customerID is an identifier, not a feature — must be excluded from ML.")
```
